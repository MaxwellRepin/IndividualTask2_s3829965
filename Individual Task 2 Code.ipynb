##### KNN

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.inspection import permutation_importance


import matplotlib.pyplot as plt
```

```python
studentperformance.head()
```

```python
studentperformance.info()
```

```python
predictor = studentperformance.drop(['Exam_Score'], axis=1)
target = studentperformance['Exam_Score']

cat_cols = ['Parental_Involvement', 
            'Access_to_Resources', 
            'Extracurricular_Activities', 
            'Motivation_Level', 
            'Internet_Access', 
            'Family_Income', 
            'Teacher_Quality', 
            'School_Type', 
            'Peer_Influence', 
            'Learning_Disabilities', 
            'Parental_Education_Level', 
            'Distance_from_Home','Gender']

num_cols = ['Hours_Studied', 
            'Attendance', 
            'Sleep_Hours', 
            'Previous_Scores', 
            'Tutoring_Sessions', 
            'Physical_Activity']


predictor_encoded = predictor.copy()
label_encoders = {}

for col in cat_cols:
    le = LabelEncoder()
    predictor_encoded[col] = le.fit_transform(predictor_encoded[col])
    label_encoders[col] = le
    

predict_train, predict_test, target_train, target_test = train_test_split(predictor_encoded, target, test_size=0.3, random_state=1)


scaler = StandardScaler()

predict_train_scaled = predict_train.copy()
predict_test_scaled = predict_test.copy()

predict_train_scaled[num_cols] = scaler.fit_transform(predict_train[num_cols])
predict_test_scaled[num_cols] = scaler.transform(predict_test[num_cols])
```

```python
mse_scores = []
rmse_scores = []
mae_scores = []

k_values = [2, 4, 6, 7, 8, 9, 10, 12, 14, 16, 18, 20]


for k in k_values:
    knn = KNeighborsRegressor(n_neighbors=k)
    knn.fit(predict_train_scaled, target_train)
    target_pred = knn.predict(predict_test_scaled)
    
    
    mse = mean_squared_error(target_test, target_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(target_test, target_pred)
    
    mse_scores.append(mse)
    rmse_scores.append(rmse)
    mae_scores.append(mae)


fig, ax = plt.subplots(figsize=(12, 10))

ax.plot(k_values, mse_scores, marker='o', label='MSE')
ax.plot(k_values, rmse_scores, marker='o', label='RMSE')
ax.plot(k_values, mae_scores, marker='o', label='MAE')
ax.set_xlabel('K-Number of Neighbors (K)', fontsize=14)
ax.set_ylabel('Error', fontsize=14)
ax.set_title('KNN Regression Metrics vs. K-Neighbors from 2 to 20 on Student Performance Dataset', fontsize=16)
ax.grid(True, alpha=0.3)
ax.legend()
plt.show()
```

```python
# k = 8

knn = KNeighborsRegressor(n_neighbors = 8)
knn.fit(predict_train_scaled, target_train)
target_pred = knn.predict(predict_test_scaled)

predictor_test_decode = predict_test.copy()

for col in cat_cols:
    le = label_encoders[col]
    predictor_test_decode[col] = le.inverse_transform(predict_test[col])
```

```python
print('KNN Regression Scores for Student Performance Dataset:')
print('='*30)

mse = mean_squared_error(target_test, target_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(target_test, target_pred)
r2 = r2_score(target_test, target_pred)

print(f'Mean Squared Error (MSE): {mse:.2f}')
print(f'Root Mean Squared Error (RMSE): {rmse:.2f}')
print(f'Mean Absolute Error (MAE): {mae:.2f}')
print(f'R-squared (R2): {r2:.2f}')
```

```python
perm_importance = permutation_importance(knn, 
                                                predict_test_scaled, 
                                                target_test, 
                                                n_repeats=15, 
                                                random_state=1)

importance_df = pd.DataFrame({
    'Feature': num_cols + cat_cols,
    'Importance': perm_importance.importances_mean
}).sort_values(by='Importance', 
               ascending=False).reset_index(drop=True)

importance_df 
```

```python
from fairlearn.metrics import MetricFrame, mean_prediction, count
import dataframe_image as dfi

metrics = {
    'N': count,
    'MSE': mean_squared_error,
    'MAE': mean_absolute_error,
    'R2': r2_score,
    'Mean Prediction': mean_prediction,
}

def fairlearn_bias_sp(target_test, target_pred, sensitive_variable, title):
    
    sp_variables = label_encoders_sp[sensitive_variable].inverse_transform(predict_test_sp[sensitive_variable])
    
    mf = MetricFrame(metrics = metrics,
                     y_true = target_test,
                     y_pred = target_pred,
                     sensitive_features = sp_variables,
                     n_boot = 100,
                     ci_quantiles = [0.05, 0.95],
                     random_state = 1
    )
    
    axes = mf.by_group.plot.bar(subplots = True, 
                                layout = [3, 2], 
                                legend = False, 
                                figsize = [16, 10],
                                title = title,
    )
    
    for ax in axes.flatten():
        ax.grid(axis = 'y',
                alpha = 0.5)

    table = mf.by_group.copy()
    table.loc['Difference'] = mf.difference()
    table.loc['Diff CI Lower'] = mf.difference_ci()[0]
    table.loc['Diff CI Upper'] = mf.difference_ci()[1]
    table.loc['Overall'] = mf.overall

    
    table['N'] = pd.to_numeric(table['N'], errors = 'coerce').round().astype('Int64')
    table.round(4).to_csv(title.replace(' ','_').replace('-', '') + '.csv')
    
    dfi.export(table.round(4), title.replace(' ','_').replace('-', '') + '_table.png')
    
    plt.savefig(title + '.png')
    plt.close()
    return mf


def fairlearn_bias(target_test, target_pred, sensitive_variable, title):
    
    variables = label_encoders[sensitive_variable].inverse_transform(predict_test[sensitive_variable])
    
    mf = MetricFrame(metrics = metrics,
                     y_true = target_test,
                     y_pred = target_pred,
                     sensitive_features = variables,
                     n_boot = 100,
                     ci_quantiles = [0.05, 0.95],
                     random_state = 1
    )
    
    axes = mf.by_group.plot.bar(subplots = True, 
                                layout = [3, 2], 
                                legend = False, 
                                figsize = [16, 10],
                                title = title,
    )
    
    for ax in axes.flatten():
        ax.grid(axis = 'y',
                alpha = 0.5)
    
    table = mf.by_group.copy()
    table.loc['Difference'] = mf.difference()
    table.loc['Diff CI Lower'] = mf.difference_ci()[0]
    table.loc['Diff CI Upper'] = mf.difference_ci()[1]
    table.loc['Overall Diff'] = mf.overall

    
    table['N'] = pd.to_numeric(table['N'], errors = 'coerce').round().astype('Int64')
    table.round(4).to_csv(title.replace(' ','_').replace('-', '') + '.csv')

    dfi.export(table.round(4), title.replace(' ','_').replace('-', '') + '_table.png')
    
    plt.savefig(title + '.png')
    plt.close()
    return mf
```

```python
def plot_disparity (mf, title, ax, ylim):
    
    diff = mf.difference().dropna()
    diff = diff.drop(['MSE', 'N'])
    
    bars = ax.bar(diff.index, 
                  diff.values, 
                  color = 'firebrick',
                  alpha = 0.7)
    
    ax.set_title(title, 
                 fontsize = 15)
    ax.set_ylabel('Disparity (max - min)')
    ax.grid(axis = 'y',
            alpha = 0.3)
    ax.set_ylim(ylim)
```

```python
fig1, axes = plt.subplots(2, 
                          3, 
                          figsize=(16, 9))

fig1.suptitle('Fairlearn Difference — Student Performance Dataset',
              fontsize=20,
              weight = 'bold',
              )

ylim = (0, 1)

plot_disparity(mf_sp_gender_knn,
               'KNN — Gender', 
               axes[0, 0],
               ylim)


plot_disparity(mf_sp_motivation_knn,
               'KNN — Motivation Level',
               axes[0, 1],
               ylim)


plot_disparity(mf_sp_parental_knn,
               'KNN — Parental Involvement',   
               axes[0, 2],
               ylim)


plot_disparity(mf_sp_gender,       
               'LassoCV — Gender',             
               axes[1, 0],
               ylim)

plot_disparity(mf_sp_motivation,   
               'LassoCV — Motivation Level',   
               axes[1, 1],
               ylim)

plot_disparity(mf_sp_parental,     
               'LassoCV — Parental Involvement', 
               axes[1, 2],
               ylim)


plt.tight_layout()
plt.savefig('difference_student_performance.png')
plt.show()
```
